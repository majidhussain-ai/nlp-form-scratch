```python
import nltk
from nltk import word_tokenize
#nltk.download('punkt')
#nltk.download('averaged_perceptron_tagger')

# Example paragraph with punctuation
paragraph = "Hello! My name is John. I love programming, especially in Python. Do you enjoy coding too?"

# Tokenize the sentence (split into words)
words = word_tokenize(paragraph)

# POS tagging
pos_tags = nltk.pos_tag(words)

# Output the tagged words
print(pos_tags)
```

```python
def edit_distance(s1, s2):
    # Get the lengths of the two strings
    m = len(s1)
    n = len(s2)

    # Create a 2D table to store results of subproblems
    dp = [[0] * (n + 1) for _ in range(m + 1)]

    # Fill the first column and first row of the table
    for i in range(m + 1):
        dp[i][0] = i  # Requires i operations to convert s1 to an empty string
    for j in range(n + 1):
        dp[0][j] = j  # Requires j operations to convert an empty string to s2

    # Fill the rest of the table
    for i in range(1, m + 1):
        for j in range(1, n + 1):
            # If characters are the same, no operation is needed
            if s1[i - 1] == s2[j - 1]:
                dp[i][j] = dp[i - 1][j - 1]
            else:
                # Take the minimum of insert, remove, or replace
                dp[i][j] = 1 + min(dp[i - 1][j],    # Remove
                                   dp[i][j - 1],    # Insert
                                   dp[i - 1][j - 1])  # Replace

    # The result will be in the bottom-right cell of the table
    return dp[m][n]

# Example Usage
s1 = "language"
s2 = "lounge"

print("Edit Distance:", edit_distance(s1, s2))
```
