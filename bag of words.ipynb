```python
# Importing necessary libraries
from collections import Counter
import re

# Function to preprocess text
def preprocess_text(text):
    # Convert text to lowercase
    text = text.lower()
    # Remove punctuation and non-alphabetic characters
    text = re.sub(r'[^a-z\s]', '', text)
    return text

# Function to create Bag of Words representation
def bag_of_words(sentences):
    # Preprocess each sentence and create a vocabulary
    processed_sentences = [preprocess_text(sentence) for sentence in sentences]
    vocabulary = set(word for sentence in processed_sentences for word in sentence.split())
    
    # Create a BoW representation for each sentence
    bow_representation = []
    for sentence in processed_sentences:
        # Count the frequency of each word in the vocabulary for the current sentence
        word_count = Counter(sentence.split())
        # Create a BoW vector based on the vocabulary
        bow_vector = {word: word_count[word] for word in vocabulary}
        bow_representation.append(bow_vector)

    return vocabulary, bow_representation

# Example sentences
sentences = [
    "Cats are great pets.",
    "Dogs are also great pets."
]

# Perform Bag of Words
vocabulary, bow_vectors = bag_of_words(sentences)

# Output the results
print("Vocabulary:", vocabulary)
print("Bag of Words Representation:")
for i, bow in enumerate(bow_vectors):
    print(f"Sentence {i + 1} BoW:", bow)
```

```python
import nltk  # Import the NLTK library for natural language processing
from nltk.tokenize import word_tokenize  # Import the tokenizer from NLTK
from collections import Counter  # Import Counter to count word frequencies
import re  # Import regular expressions for text cleaning
import pandas as pd  # Import pandas for creating DataFrames

# Download the punkt tokenizer if not already downloaded
#nltk.download('punkt')

# Function to preprocess the text
def preprocess_text(text):
    # Convert text to lowercase to standardize it
    text = text.lower()
    # Remove punctuation and non-alphabetic characters
    text = re.sub(r'[^a-z\s]', '', text)
    return text  # Return the cleaned text

# Function to create a Bag of Words representation
def bag_of_words(paragraph):
    # Preprocess the paragraph
    processed_paragraph = preprocess_text(paragraph)
    # Tokenize the processed paragraph into individual words
    tokenized_words = word_tokenize(processed_paragraph)
    
    # Create a frequency distribution of words using Counter
    word_count = Counter(tokenized_words)
    
    return word_count  # Return the word frequency count

# Example paragraph
paragraph = """
Natural language processing (NLP) is a fascinating field of artificial intelligence that focuses on the interaction between computers and humans through natural language.
It enables computers to understand, interpret, and respond to human language in a valuable way.
"""

# Apply Bag of Words to the example paragraph
bow_representation = bag_of_words(paragraph)

# Create a DataFrame from the Bag of Words representation
bow_df = pd.DataFrame(bow_representation.items(), columns=['Word', 'Frequency'])

# Display the DataFrame as a table
print("Bag of Words Representation:")
print(bow_df)
```
